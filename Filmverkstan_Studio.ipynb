# 🎬 Filmverkstan Studio

Din egen gratis AI-filmstudio på Googles grafikkort: **bilder, rörliga scener, pratande personer, svensk röst, musik och färdiga filmer** – med knappar, utan kod.

**Så här gör du**
1. Kontrollera att grafikkortet är på: **Körning → Ändra körningstyp → T4 GPU → Spara**.
2. Tryck ▶ på rutan nedan. Första gången tar det ca 5 minuter.
3. Klicka på länken som slutar på **gradio.live** när den dyker upp. Där finns hela studion.

💡 Låt den här fliken vara öppen medan du jobbar. Gratis-Colab har en tidsgräns per dag.

```python
#@title ▶ Starta Filmverkstan Studio { display-mode: "form" }
#@markdown **Spara det du skapar i Google Drive** (rekommenderas – annars försvinner det när Colab stängs):
Spara_resultat_i_Google_Drive = True #@param {type:"boolean"}
#@markdown **Spara AI-modellerna i Google Drive** – gör att det startar på några minuter nästa gång, men kräver ca 30 GB ledigt i din Drive:
Spara_AI_modeller_i_Google_Drive = False #@param {type:"boolean"}
import os, subprocess, sys
import torch
if not torch.cuda.is_available():
    raise SystemExit("❌ Inget grafikkort! Körning → Ändra körningstyp → T4 GPU → Spara. Tryck sedan ▶ igen.")
if Spara_resultat_i_Google_Drive or Spara_AI_modeller_i_Google_Drive:
    from google.colab import drive
    drive.mount("/content/drive")
os.environ["FV_DRIVE_MODELS"] = "1" if Spara_AI_modeller_i_Google_Drive else "0"
R = "/content/filmverkstan"
if os.path.exists(R):
    subprocess.run(["git", "-C", R, "pull", "-q"])
else:
    subprocess.run(["git", "clone", "-q", "https://github.com/fredriklofsten/filmverkstan.git", R], check=True)
subprocess.run([sys.executable, f"{R}/studio/setup_env.py"], check=True)
print("\n🚀 Startar studion… klicka på länken som slutar på gradio.live när den visas nedan.\n")
!cd {R}/studio && python app.py
```
