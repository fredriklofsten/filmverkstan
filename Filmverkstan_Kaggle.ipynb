# 🎬 Filmverkstan Studio – Kaggle-versionen

Samma studio som i Colab, men på Kaggles gratis grafikkort (**2 × T4, ca 30 timmar i veckan**).

**Första gången (en gång):**
1. Högermenyn → **Session options** → **Accelerator: GPU T4 x2**.
2. Samma meny → **Internet: On** (kräver att ditt telefonnummer är verifierat på Kaggle).

**Varje gång:**
1. Tryck ▶ på rutan nedan (eller **Run All**). Första gången tar det ca 10 minuter.
2. Klicka på länken som slutar på **gradio.live**. Där finns hela studion.
3. Färdiga filer hamnar under **Output → resultat** till höger. Ladda ner det du vill spara.

💡 Stäng av sessionen (⏻ uppe till höger) när du är klar, så sparar du dina GPU-timmar.

```python
# ▶ Starta Filmverkstan Studio
import os, subprocess, sys, torch
if not torch.cuda.is_available():
    raise SystemExit("❌ Inget grafikkort! Session options → Accelerator → GPU T4 x2. Tryck sedan ▶ igen.")
print(f"✅ {torch.cuda.device_count()} grafikkort: {torch.cuda.get_device_name(0)}")
try:
    subprocess.run(["git", "ls-remote", "https://github.com/fredriklofsten/filmverkstan.git"], check=True, capture_output=True, timeout=30)
except Exception:
    raise SystemExit("❌ Ingen internetanslutning! Session options → Internet → On (kräver verifierat telefonnummer).")
os.makedirs("/content", exist_ok=True)
R = "/content/filmverkstan"
if os.path.exists(R):
    subprocess.run(["git", "-C", R, "pull", "-q"])
else:
    subprocess.run(["git", "clone", "-q", "https://github.com/fredriklofsten/filmverkstan.git", R], check=True)
subprocess.run([sys.executable, f"{R}/studio/setup_env.py"], check=True)
print("\n🚀 Startar studion… klicka på länken som slutar på gradio.live när den visas nedan.\n")
!cd {R}/studio && python app.py
```
